```python
##q1 whats the total revenue generated by male vs female
def run_sql(query):
  return pd.read_sql_query(query, conn)
```

```python
run_sql ("""
Select gender, sum("purchase_amount_(usd)") as revenue
from customer
group by gender
""")
```

```python
#q2 which customer used a discount but still spent more than average purchase amount
run_sql("""
select customer_id, "purchase_amount_(usd)"
from customer
where discount_applied = 'Yes' and "purchase_amount_(usd)" >= (select avg("purchase_amount_(usd)") from customer)
""")
```

```python
##q3) which are the top 5 best products with the highest average review rating ?
run_sql("""
Select item_purchased , Round(avg("review_rating"),2) as avg_rating
from customer
group by item_purchased
order by avg_rating desc
limit 5
""")
```

```python
##q4) compare the average product amounts between Standard and Express shipping ?
## if we find out express shipping sells more then its a clear indication that they should
##invest more in fast erdeliverey options
run_sql("""
Select shipping_type,
ROund(Avg("purchase_amount_(usd)"),2) from customer
where shipping_type in ("Standard","Express")
group by shipping_type
""")
#hence business should focus on express faster delievery options
```

```python
### Q5) do subscriberd customers spend more ? Compare average spend and total revenue
## between subscribers and non subscribers
 ## the results will tell us whether subscriptions do work or no
run_sql("""
 Select subscription_status ,
 Count(customer_id) as total_customers,
 ROund(Avg("purchase_amount_(usd)"),2) as avg_spend,
 sum("purchase_amount_(usd)") as total_revenue
 from customer
 group by subscription_status
 order by total_revenue,avg_spend desc
""" )
```

```python
##Q6 which products have the highest percentage of purchases with discounts applied
run_sql("""
Select item_purchased,
Round(100.0 * sum(case when discount_applied = "Yes" THEN 1 ELSE 0 END) / count(*), 2) as discount_rate
from customer
group by item_purchased
order by discount_rate desc
limit 5
""")
```

```python
##Q7 important question :Segment customers into neew ,Returning  and loyal
## based on their total number of previous purchases
##and show the count of each segment ?
run_sql("""
with customer_type as (
select customer_id,previous_purchases ,
Case
  when previous_purchases = 1 THEN "New"
  when previous_purchases BETWEEN 2 AND 10 THEN "Returning"
  ELSE "Loyal"
  End AS customer_segment
  from customer
  )
  select customer_segment,count(*) as "Number of Customers"
  from customer_type
  group by customer_segment
  order by "Number of Customers" desc
""")
```

```python
##q8) solve again   what are the top 3 most purchased products within each category ?
run_sql("""
  with item_counts as (
    select
      category,
      item_purchased,
      count(customer_id) as total_orders
    from customer
    group by category, item_purchased
  )
  select
    ic1.category,
    ic1.item_purchased,
    ic1.total_orders
  from item_counts ic1
  where (
    select count(*)
    from item_counts ic2
    where ic2.category = ic1.category
      and ic2.total_orders > ic1.total_orders
  ) < 3
  order by ic1.category, ic1.total_orders DESC, ic1.item_purchased;
""")
```

```python
##Q9 Are customers who are repeat buyers (more than 5 previous purchases )
## also likely to subscribe?
run_sql("""
select subscription_status,
count(customer_id) as repeat_buyers
from customer
where previous_purchases > 5
group by subscription_status order by repeat_buyers desc
 """)
```

```python
##q10 what is the revenue contribution of each age group?
run_sql("""
select age_group,
round(sum("purchase_amount_(usd)"),2) as total_revenue
from customer
group by age_group
order by total_revenue desc
""")
```
